#### Decorators
Decorators are a powerful and flexible feature in Python that allows you to modify the behaviour of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

```python
# function copy
# closures
# decorators
```

```python
# function copy
def welcome():
    return "Welcome to the advanced python course"

welcome()
```

```python
wel = welcome
print(wel())
del welcome
print(wel())
```

```python
# closures functions

def main_welcome(msg):
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        print(msg)
        
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome("Hey there welcome")
```

```python
def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        func("someone welcomes you all")
        
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(print)
```

```python
def main_welcome(func, lst):
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        print(func(lst))
        
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(len, [1,2,3,4,5,6])
```

```python
# decorator

def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        func()
        
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
def course_introduction():
    print("This is an advanced Python course")
    
course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
#### decorator

@main_welcome
def course_introduction():
    print("This is an advanced Python course")
```

```python
## Decorator

def my_decorator(func):
    def wrapper():
        print("Something is happenning before the function is called")
        func()
        print("Something is happenning after the function is called")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hehe")
```

```python
say_hello()
```

```python
# Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("Hehe")
```

```python
say_hello()
```
