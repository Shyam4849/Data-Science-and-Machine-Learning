#### Data Manipulation and Analysis with Pandas
Data Manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform, and extract insghts from data. In this lesson, we will cover various data manipulation and analysis techniques using Pandas.

```python
import pandas as pd
```

```python
df = pd.read_csv('sales_practice_dataset.csv')

# fetch first 5 rows
df.head(5) 
```

```python
# Handling Missing Values
df.isnull().any()
```

```python
df.isnull().sum()
```

```python
df_filled = df.fillna(0)
```

```python
# filling missing values with the mean of the column

df['Sales_fillNA'] = df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df.dtypes
```

```python
# Renaming columns
df = df.rename(columns={'Date': 'Sales Date'})
df.head()
```

```python
# Change datatypes
df['Value_new'] = df['Value'].fillna(df['Value'].mean()).astype(int)
df.head()
```

```python
df['New Value'] = df['Value'].apply(lambda x: x*2)
df.head()
```

```python
# Data Aggregating and Grouping
df.head()
```

```python
grouped_mean = df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['Product', 'Region'])['Value'].sum()
print(grouped_sum)
```

```python
grouped_sum = df.groupby(['Product', 'Region'])['Value'].mean()
print(grouped_sum)
```

```python
# Aggregate multiple functions
grouped_agg = df.groupby('Region')['Value'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
# Merging and Joining DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
# Merge DataFrame on the 'Key Columns'
pd.merge(df1, df2, on="Key",how="inner")
```

```python
pd.merge(df1, df2, on="Key",how="outer")
```

```python
pd.merge(df1, df2, on="Key",how="left")
```

```python
pd.merge(df1, df2, on="Key",how="right")
```
