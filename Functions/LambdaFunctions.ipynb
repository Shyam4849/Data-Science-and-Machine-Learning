### Lambda Function
Lambda function are small anonymous functions defined using *__lambda__* keyword.They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.

```python
# Syntax
lambda argumments: expression
```

```python
def addition(a,b):
    return a+b
```

```python
addition(2,4)
```

```python
addition = lambda a,b: a+b
type(addition)
print(addition(2,5))
```

```python
def even(num):
    if num%2 == 0:
        return True

even(24)
```

```python
even1 = lambda num: num%2 == 0
print(even1(12))
```

```python
def addition(x,y,z):
    return x+y+z

addition(12,3,4)
```

```python
addition1 = lambda x,y,z:x+y+z
addition1(1,2,3)
```

```python
# map() -> applies a function to all items in a list
numbers=[1,2,3,4,5,6]

def square(number):
    return number ** 2

square(2)
```

```python
list(map(lambda x:x**2,numbers))
```
